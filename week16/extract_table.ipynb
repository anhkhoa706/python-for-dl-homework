```python
import json
import pandas as pd

# === Load and parse the JSON files ===

# Replace these with your actual file paths
file_illegal = "/home/wins057/Documents/Projects/python-for-dl-homework/week16/violation_results_illegal_parking_20250606_022514.json"
file_no_violation = "/home/wins057/Documents/Projects/python-for-dl-homework/week16/violation_results_no_violations_20250606_023659.json"

# Load illegal parking JSON
with open(file_illegal, 'r') as f:
    data_illegal = json.load(f)

# Load no violation JSON
with open(file_no_violation, 'r') as f:
    data_no_violation = json.load(f)

# Extract video filename and has_violation
df_illegal = pd.DataFrame([
    {
        "Video Filename": vid["filename"],
        "Has Violation": vid["has_violation"]
    }
    for vid in data_illegal["videos"]
])

df_no_violation = pd.DataFrame([
    {
        "Video Filename": vid["filename"],
        "Has Violation": vid["has_violation"]
    }
    for vid in data_no_violation["videos"]
])

# Display both
print("=== Illegal Parking JSON ===")
print(df_illegal)

print("\n=== No Violation JSON ===")
print(df_no_violation)
```

```python
df_illegal.to_csv("illegal_parking_results.csv", index=False)
df_no_violation.to_csv("no_violation_results.csv", index=False)
```
