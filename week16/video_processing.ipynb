```python
import cv2
import os

video_root = "/home/wins057/Documents/Projects/python-for-dl-homework/week16/dataset/video"
image_root = "/home/wins057/Documents/Projects/python-for-dl-homework/week16/dataset/image"

# Traverse all subfolders under dataset/video
for subfolder in os.listdir(video_root):
    subfolder_path = os.path.join(video_root, subfolder)
    if not os.path.isdir(subfolder_path):
        continue

    # Create matching output subfolder
    output_subfolder = os.path.join(image_root, subfolder)
    os.makedirs(output_subfolder, exist_ok=True)

    # Process each .mp4 video in the subfolder
    for video_file in os.listdir(subfolder_path):
        if not video_file.endswith(".mp4"):
            continue

        video_path = os.path.join(subfolder_path, video_file)
        video_name = os.path.splitext(video_file)[0]

        output_video_folder = os.path.join(output_subfolder, video_name)
        os.makedirs(output_video_folder, exist_ok=True)

        cap = cv2.VideoCapture(video_path)
        if not cap.isOpened():
            print(f"❌ Failed to open {video_file}")
            continue

        frame_id = 0
        saved_id = 0

        while True:
            ret, frame = cap.read()
            if not ret:
                break

            # Save every 2nd frame (reduce 30 fps → ~15 fps)
            if frame_id % 2 == 0:
                output_frame_path = os.path.join(output_video_folder, f"frame_{saved_id:05d}.jpg")
                cv2.imwrite(output_frame_path, frame)
                saved_id += 1

            frame_id += 1

        cap.release()
        print(f"✅ Saved {saved_id} frames from {video_file} to {output_video_folder}")

print("🎉 All videos processed.")
```
