```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Define the sigmoid function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Define tanh function
def tanh(x):
    return np.tanh(x)

# Define ReLU function
def relu(x):
    return np.maximum(0, x)
```

```python
# Generate an array of x values from -10 to 10
x = np.linspace(-10., 10., 100)
```

```python
# Calculate corresponding y values using the sigmoid function
y = sigmoid(x)

# Plot the sigmoid function
plt.figure(figsize=(8, 6))
plt.plot(x, y, label="Sigmoid Function", color='b')
plt.title('Sigmoid Function')
plt.xlabel('x')
plt.ylabel('Sigmoid(x)')
plt.grid(True)
plt.legend()
plt.show()
```

```python
# Calculate corresponding y values using the ReLU function
y = relu(x)

# Plot the sigmoid function
plt.figure(figsize=(8, 6))
plt.plot(x, y, label="ReLu Function", color='b')
plt.title('ReLU Function')
plt.xlabel('x')
plt.ylabel('ReLU(x)')
plt.grid(True)
plt.legend()
plt.show()
```

```python
# Calculate corresponding y values using the tanh function
y = tanh(x)

# Plot the sigmoid function
plt.figure(figsize=(8, 6))
plt.plot(x, y, label="Tanh Function", color='b')
plt.title('Tanh Function')
plt.xlabel('x')
plt.ylabel('Tanh(x)')
plt.grid(True)
plt.legend()
plt.show()
```
